EJERCICIO 1 CALCULAR IMPUESTO A LAS ACTIVIDADES ECONOMICAS

```python
def calcular_impuesto(monto):
    # Tabla de rangos:
    tabla = [
        (0.01, 500.00, 1.5, 0.0),
        (500.01, 1000.00, 1.5, 3.0),
        (1000.01, 2000.00, 3.0, 3.0),
        (2000.01, 3000.00, 6.0, 3.0),
        (3000.01, 6000.00, 9.0, 2.0),
        (6000.01, 18000.00, 15.0, 2.0),
        (18000.01, 30000.00, 39.0, 2.0),
        (30000.01, 60000.00, 63.0, 1.0),
        (60000.01, 100000.00, 93.0, 0.8),
        (100000.01, 200000.00, 125.0, 0.7),
        (200000.01, 300000.00, 195.0, 0.6),
        (300000.01, 400000.00, 255.0, 0.45),
        (400000.01, 500000.00, 300.0, 0.4),
        (500000.01, 1000000.00, 340.0, 0.30),
        (1000000.01, 99999999.00, 490.0, 0.18)
    ]

    for desde, hasta, precio, adicional in tabla:
        if desde <= monto <= hasta:
            diferencia = monto - desde
            impuesto = (diferencia / 1000.0) * adicional + precio
            return round(impuesto, 2)

    return 0.0

# Entrada del usuario
try:
    monto_actividad = float(input("Ingrese el monto de la actividad económica: $"))
    total_pagar = calcular_impuesto(monto_actividad)
    print(f"El valor del impuesto a pagar es: ${total_pagar:.2f}")
except ValueError:
    print("Por favor, ingrese un monto numérico válido.")
```

```python
# Factores de conversión referenciados a Metro Cuadrado (m²)
FACTORES_A_METROS2 = {
    "1": ("Pie Cuadrado", 0.0929),
    "2": ("Vara Cuadrada", 0.698737),
    "3": ("Yarda Cuadrada", 0.9144),
    "4": ("Metro Cuadrado", 1.43),
    "5": ("Tarea", 628.86),
    "6": ("Manzana", 7000),
    "7": ("Hectárea", 1.429)
}

def menu_unidades():
    print("\nUnidades disponibles:")
    for clave, (nombre, _) in FACTORES_A_METROS2.items():
        print(f"{clave}. {nombre}")

def convertir_area(valor, origen_key, destino_key):
    # Convertir primero de la unidad origen a Metros Cuadrados
    factor_origen = FACTORES_A_METROS2[origen_key][1]
    valor_en_m2 = valor * factor_origen

    # Convertir de Metros Cuadrados a la unidad destino
    factor_destino = FACTORES_A_METROS2[destino_key][1]
    resultado = valor_en_m2 / factor_destino

    return resultado

def ejecutar_conversor():
    print("=== CONVERSOR DE ÁREA (EL SALVADOR) ===")
    menu_unidades()

    opc_origen = input("\nSeleccione la unidad de origen (1-7): ")
    opc_destino = input("Seleccione la unidad a la que desea convertir (1-7): ")

    if opc_origen in FACTORES_A_METROS2 and opc_destino in FACTORES_A_METROS2:
        try:
            valor = float(input(f"Ingrese la cantidad en {FACTORES_A_METROS2[opc_origen][0]}: "))
            res = convertir_area(valor, opc_origen, opc_destino)

            nombre_origen = FACTORES_A_METROS2[opc_origen][0]
            nombre_destino = FACTORES_A_METROS2[opc_destino][0]

            print(f"\n---> {valor} {nombre_origen} equivale a: {res:.6f} {nombre_destino}")
        except ValueError:
            print("Error: Ingrese un valor numérico válido.")
    else:
        print("Opción no válida.")

# Ejecutar el conversor
ejecutar_conversor()
```

```python
import json
import math
from collections import Counter
from functools import reduce
# ==========================================
# Explicacion
# ==========================================
"""
Diferencia entre los enfoques de programación:

- Sistema Académico usando JSON / Diccionarios: Permite almacenar y estructurar
  datos en formato clave-valor de forma simple, liviana y portable para intercambio de información.
- Usando POO (Programación Orientada a Objetos): Modela entidades mediante clases y objetos,
  permitiendo agrupar datos (atributos) y comportamientos (métodos) de manera organizada y reutilizable.
- Funciones de usuario y predefinidas: Las funciones predefinidas son utilidades nativas del
  lenguaje (ej. sum(), len(), max()), mientras que las de usuario son bloques personalizados para tareas específicas.
- Estructuras de control, listas y diccionarios: Las listas y diccionarios ordenan y agrupan
  la información, mientras que las estructuras de control (if, for, while) toman decisiones y recorren los datos.
"""
# ==========================================
# CLASE ESTADÍSTICA (CÁLCULOS)
# ==========================================

class Estadistica:
    @staticmethod
    def media(datos):
        if not datos:
            return 0.0
        return sum(datos) / len(datos)

    @staticmethod
    def mediana(datos):
        if not datos:
            return 0.0
        datos_ordenados = sorted(datos)
        n = len(datos_ordenados)
        mitad = n // 2
        if n % 2 == 0:
            return (datos_ordenados[mitad - 1] + datos_ordenados[mitad]) / 2.0
        return float(datos_ordenados[mitad])

    @staticmethod
    def moda(datos):
        if not datos:
            return []
        conteo = Counter(datos)
        max_frecuencia = max(conteo.values())
        return [k for k, v in conteo.items() if v == max_frecuencia]

    @staticmethod
    def varianza(datos):
        if len(datos) < 2:
            return 0.0
        m = Estadistica.media(datos)
        return sum((x - m) ** 2 for x in datos) / (len(datos) - 1)

    @staticmethod
    def desviacion(datos):
        return math.sqrt(Estadistica.varianza(datos))

    @staticmethod
    def rango(datos):
        if not datos:
            return 0
        return max(datos) - min(datos)

    @staticmethod
    def frecuencia(datos):
        return dict(Counter(datos))


# ==========================================
# CLASE ESTUDIANTE
# ==========================================

class Estudiante:
    def __init__(self, nombre, edad, tiempo_traslado_min, horas_uso_mvil, notas):
        self.nombre = nombre
        self.edad = edad
        self.tiempo_traslado_min = tiempo_traslado_min
        self.horas_uso_mvil = horas_uso_mvil
        self.notas = notas  # Diccionario: {"Materia": nota, ...}

    def promedio_general(self):
        val_notas = list(self.notas.values())
        if not val_notas:
            return 0.0
        return sum(val_notas) / len(val_notas)

    def aprobo(self, nota_minima=6.0):
        return self.promedio_general() >= nota_minima

    def to_dict(self):
        return {
            "nombre": self.nombre,
            "edad": self.edad,
            "tiempo_traslado_min": self.tiempo_traslado_min,
            "horas_uso_mvil": self.horas_uso_mvil,
            "notas": self.notas,
            "promedio_general": round(self.promedio_general(), 2),
            "aprobado": self.aprobo()
        }


# ==========================================
#CLASE SISTEMA ACADÉMICO
# ==========================================

class SistemaAcademico:
    def __init__(self):
        self.estudiantes = []
        self.estadistica = Estadistica()

    def cargar_estudiantes(self, datos):
        self.estudiantes = list(map(
            lambda dato: Estudiante(
                dato["nombre"],
                dato["edad"],
                dato["tiempo_traslado_min"],
                dato["horas_uso_mvil"],
                dato["notas"]
            ),
            datos
        ))

    def reporte_general(self):
        edades = list(map(lambda e: e.edad, self.estudiantes))
        traslado = list(map(lambda e: e.tiempo_traslado_min, self.estudiantes))
        mvil = list(map(lambda e: e.horas_uso_mvil, self.estudiantes))

        print("\n=== REPORTE GENERAL DE VARIABLES GENERALES ===")
        print(f"Edad                -> Media: {self.estadistica.media(edades):.2f} | Desviación: {self.estadistica.desviacion(edades):.2f}")
        print(f"Tiempo Traslado(min)-> Media: {self.estadistica.media(traslado):.2f} | Desviación: {self.estadistica.desviacion(traslado):.2f}")
        print(f"Horas Uso Móvil     -> Media: {self.estadistica.media(mvil):.2f} | Desviación: {self.estadistica.desviacion(mvil):.2f}")

    def reporte_campo(self, campo):
        valores = list(filter(lambda x: x is not None, map(lambda e: getattr(e, campo, None), self.estudiantes)))

        if valores:
            print(f"\n=== ANÁLISIS DETALLADO DE: {campo.upper()} ===")
            print(f"Media Aritmética  : {self.estadistica.media(valores):.2f}")
            print(f"Mediana           : {self.estadistica.mediana(valores):.2f}")
            print(f"Moda              : {self.estadistica.moda(valores)}")
            print(f"Varianza          : {self.estadistica.varianza(valores):.2f}")
            print(f"Desviación Estándar: {self.estadistica.desviacion(valores):.2f}")
            print(f"Rango             : {self.estadistica.rango(valores):.2f}")
            print(f"Frecuencia        : {self.estadistica.frecuencia(valores)}")
        else:
            print("Campo no encontrado.")

    def resumen_por_materia(self):
        todas_materias = list(map(lambda e: list(e.notas.keys()), self.estudiantes))
        materias_unicas = list(set(reduce(lambda x, y: x + y, todas_materias, [])))

        print("\n=== ANÁLISIS ESTADÍSTICO DETALLADO POR MATERIA ===")
        for materia in materias_unicas:
            notas = list(filter(lambda x: x is not None, map(lambda e: e.notas.get(materia), self.estudiantes)))

            if notas:
                print(f"\n--------------------------------------------------")
                print(f"MATERIA: {materia.upper()}")
                print(f"--------------------------------------------------")
                print(f"  Media             : {self.estadistica.media(notas):.2f}")
                print(f"  Mediana           : {self.estadistica.mediana(notas):.2f}")
                print(f"  Moda              : {self.estadistica.moda(notas)}")
                print(f"  Varianza          : {self.estadistica.varianza(notas):.2f}")
                print(f"  Desviación Estándar: {self.estadistica.desviacion(notas):.2f}")
                print(f"  Rango             : {self.estadistica.rango(notas):.2f}")

    def ver_aprobados_reprobados(self, nota_minima=6.0):
        aprobados = list(filter(lambda e: e.aprobo(nota_minima), self.estudiantes))
        reprobados = list(filter(lambda e: not e.aprobo(nota_minima), self.estudiantes))

        print("\n" + "="*50)
        print("RESUMEN DE APROBADOS Y REPROBADOS")
        print("="*50)

        print(f"\nAPROBADOS ({len(aprobados)} estudiantes):")
        for e in aprobados:
            print(f"  - {e.nombre}: {e.promedio_general():.2f}")

        print(f"\nREPROBADOS ({len(reprobados)} estudiantes):")
        if reprobados:
            for e in reprobados:
                print(f"  - {e.nombre}: {e.promedio_general():.2f}")
        else:
            print("  Ningún estudiante reprobó.")

        total = len(self.estudiantes)
        if total > 0:
            tasa = (len(aprobados) / total) * 100
            print(f"\nTasa de aprobación: {tasa:.1f}%")

    def mostrar_estudiantes(self):
        print("\n=== LISTA COMPLETA DE ESTUDIANTES EN JSON ===")
        print("="*80)
        estudiantes_dict = list(map(lambda e: e.to_dict(), self.estudiantes))
        for estudiante in estudiantes_dict:
            print(json.dumps(estudiante, ensure_ascii=False, indent=2))
            print("-"*80)

    def promedios_por_estudiante(self):
        print("\n=== PROMEDIO GENERAL POR ESTUDIANTE ===")
        for e in self.estudiantes:
            print(f"\n{e.nombre}")
            print("-" * len(e.nombre))
            for mat, nota in e.notas.items():
                print(f"  {mat:<42}: {nota:.2f}")
            print(f"  {'Promedio general':<42}: {e.promedio_general():.2f}")

    def promedio_general_grupo(self):
        promedios = list(map(lambda e: e.promedio_general(), self.estudiantes))
        promedio_grupo = reduce(lambda x, y: x + y, promedios, 0) / len(promedios) if promedios else 0
        print(f"\nPromedio general del grupo: {promedio_grupo:.2f}")


# ==========================================
# DATASET 20 ESTUDIANTES
# ==========================================

datos_estudiantes = [
    {
        "nombre": "Samuel Alexander Portillo Ceron",
        "edad": 20,
        "tiempo_traslado_min": 80,
        "horas_uso_mvil": 6.0,
        "notas": {
            "Administración de Base de Datos II": 10.0,
            "Estadística Computacional": 10.0,
            "Programación III": 9.0,
            "Estructura y Diseño de Circuitos Digitales": 10.0,
            "Inglés": 10.0
        }
    },
    {
        "nombre": "Andrea Beatriz Chavez Rivas",
        "edad": 20,
        "tiempo_traslado_min": 25,
        "horas_uso_mvil": 4.5,
        "notas": {
            "Administración de Base de Datos II": 7.0,
            "Estadística Computacional": 7.0,
            "Programación III": 6.5,
            "Estructura y Diseño de Circuitos Digitales": 7.5,
            "Inglés": 8.0
        }
    },
    {
        "nombre": "Carlos David Turcios Rivas",
        "edad": 20,
        "tiempo_traslado_min": 25,
        "horas_uso_mvil": 5.0,
        "notas": {
            "Administración de Base de Datos II": 10.0,
            "Estadística Computacional": 9.0,
            "Programación III": 8.5,
            "Estructura y Diseño de Circuitos Digitales": 9.0,
            "Inglés": 10.0
        }
    },
    {
        "nombre": "Oscar Ivan Mejia Portillo",
        "edad": 23,
        "tiempo_traslado_min": 40,
        "horas_uso_mvil": 7.0,
        "notas": {
            "Administración de Base de Datos II": 6.0,
            "Estadística Computacional": 6.0,
            "Programación III": 5.5,
            "Estructura y Diseño de Circuitos Digitales": 6.5,
            "Inglés": 7.0
        }
    },
    {
        "nombre": "Gilda Mileidy González Castro",
        "edad": 19,
        "tiempo_traslado_min": 30,
        "horas_uso_mvil": 5.0,
        "notas": {
            "Administración de Base de Datos II": 10.0,
            "Estadística Computacional": 7.5,
            "Programación III": 9.0,
            "Estructura y Diseño de Circuitos Digitales": 10.0,
            "Inglés": 10.0
        }
    },
    {
        "nombre": "Sofia Nicole Aguilar Reyes",
        "edad": 18,
        "tiempo_traslado_min": 10,
        "horas_uso_mvil": 3.0,
        "notas": {
            "Administración de Base de Datos II": 9.5,
            "Estadística Computacional": 9.0,
            "Programación III": 9.0,
            "Estructura y Diseño de Circuitos Digitales": 9.5,
            "Inglés": 10.0
        }
    },
    {
        "nombre": "Kimberly Michelle Castro Zelaya",
        "edad": 18,
        "tiempo_traslado_min": 20,
        "horas_uso_mvil": 9.0,
        "notas": {
            "Administración de Base de Datos II": 8.5,
            "Estadística Computacional": 9.0,
            "Programación III": 10.0,
            "Estructura y Diseño de Circuitos Digitales": 8.0,
            "Inglés": 9.5
        }
    },
    {
        "nombre": "Luis Fernando Escobar Molina",
        "edad": 21,
        "tiempo_traslado_min": 35,
        "horas_uso_mvil": 6.5,
        "notas": {
            "Administración de Base de Datos II": 7.5,
            "Estadística Computacional": 8.5,
            "Programación III": 8.0,
            "Estructura y Diseño de Circuitos Digitales": 7.0,
            "Inglés": 6.0
        }
    },
    {
        "nombre": "Yeimy Julissa Parada Chávez",
        "edad": 20,
        "tiempo_traslado_min": 45,
        "horas_uso_mvil": 3.0,
        "notas": {
            "Administración de Base de Datos II": 9.0,
            "Estadística Computacional": 9.5,
            "Programación III": 9.0,
            "Estructura y Diseño de Circuitos Digitales": 9.0,
            "Inglés": 10.0
        }
    },
    {
        "nombre": "Karen Michelle Guardado Lopez",
        "edad": 19,
        "tiempo_traslado_min": 20,
        "horas_uso_mvil": 5.0,
        "notas": {
            "Administración de Base de Datos II": 8.5,
            "Estadística Computacional": 8.5,
            "Programación III": 7.5,
            "Estructura y Diseño de Circuitos Digitales": 8.0,
            "Inglés": 9.0
        }
    },
    {
        "nombre": "Saúl Alexander Jimenez Garcia",
        "edad": 21,
        "tiempo_traslado_min": 60,
        "horas_uso_mvil": 5.0,
        "notas": {
            "Administración de Base de Datos II": 8.5,
            "Estadística Computacional": 8.0,
            "Programación III": 10.0,
            "Estructura y Diseño de Circuitos Digitales": 8.5,
            "Inglés": 9.0
        }
    },
    {
        "nombre": "Bryan Alexander Hernandez Cruz",
        "edad": 24,
        "tiempo_traslado_min": 55,
        "horas_uso_mvil": 8.0,
        "notas": {
            "Administración de Base de Datos II": 5.0,
            "Estadística Computacional": 5.0,
            "Programación III": 5.5,
            "Estructura y Diseño de Circuitos Digitales": 6.0,
            "Inglés": 6.0
        }
    },
    {
        "nombre": "Miguel Alejandro Gómez Saravia",
        "edad": 19,
        "tiempo_traslado_min": 15,
        "horas_uso_mvil": 4.0,
        "notas": {
            "Administración de Base de Datos II": 9.0,
            "Estadística Computacional": 10.0,
            "Programación III": 9.0,
            "Estructura y Diseño de Circuitos Digitales": 9.0,
            "Inglés": 9.5
        }
    },
    {
        "nombre": "Daniela Alejandra Ventura Sosa",
        "edad": 20,
        "tiempo_traslado_min": 18,
        "horas_uso_mvil": 4.0,
        "notas": {
            "Administración de Base de Datos II": 9.0,
            "Estadística Computacional": 9.5,
            "Programación III": 9.5,
            "Estructura y Diseño de Circuitos Digitales": 9.0,
            "Inglés": 8.5
        }
    },
    {
        "nombre": "Josué Stiven Funes Benavides",
        "edad": 19,
        "tiempo_traslado_min": 30,
        "horas_uso_mvil": 5.0,
        "notas": {
            "Administración de Base de Datos II": 9.0,
            "Estadística Computacional": 9.0,
            "Programación III": 9.5,
            "Estructura y Diseño de Circuitos Digitales": 9.0,
            "Inglés": 10.0
        }
    },
    {
        "nombre": "Marvin Ernesto Alvarado Diaz",
        "edad": 22,
        "tiempo_traslado_min": 12,
        "horas_uso_mvil": 2.5,
        "notas": {
            "Administración de Base de Datos II": 7.0,
            "Estadística Computacional": 7.0,
            "Programación III": 8.0,
            "Estructura y Diseño de Circuitos Digitales": 6.5,
            "Inglés": 7.5
        }
    },
    {
        "nombre": "Jefferson Steven Castillo Leiva",
        "edad": 19,
        "tiempo_traslado_min": 15,
        "horas_uso_mvil": 2.3,
        "notas": {
            "Administración de Base de Datos II": 9.0,
            "Estadística Computacional": 9.0,
            "Programación III": 9.0,
            "Estructura y Diseño de Circuitos Digitales": 9.5,
            "Inglés": 10.0
        }
    },
    {
        "nombre": "Ramael Ernesto Cordova Colindres",
        "edad": 22,
        "tiempo_traslado_min": 7,
        "horas_uso_mvil": 13.52,
        "notas": {
            "Administración de Base de Datos II": 9.0,
            "Estadística Computacional": 9.0,
            "Programación III": 9.5,
            "Estructura y Diseño de Circuitos Digitales": 9.0,
            "Inglés": 10.0
        }
    },
    {
        "nombre": "Carlos Daniel Parada Avalos",
        "edad": 25,
        "tiempo_traslado_min": 10,
        "horas_uso_mvil": 5.52,
        "notas": {
            "Administración de Base de Datos II": 8.0,
            "Estadística Computacional": 8.5,
            "Programación III": 8.0,
            "Estructura y Diseño de Circuitos Digitales": 8.0,
            "Inglés": 9.0
        }
    },
    {
        "nombre": "Daniel Enrique Quintanilla Paniagua",
        "edad": 19,
        "tiempo_traslado_min": 15,
        "horas_uso_mvil": 4.0,
        "notas": {
            "Administración de Base de Datos II": 9.0,
            "Estadística Computacional": 9.5,
            "Programación III": 9.0,
            "Estructura y Diseño de Circuitos Digitales": 10.0,
            "Inglés": 10.0
        }
    }
]


# ==========================================
# MENÚ PRINCIPAL
# ==========================================

def menu():
    sistema = SistemaAcademico()
    sistema.cargar_estudiantes(datos_estudiantes)

    while True:
        print("\n========== SISTEMA ESTADÍSTICO ACADÉMICO ==========")
        print("1. Reporte general (edad, tiempo traslado, horas uso móvil)")
        print("2. Reporte de una variable específica")
        print("3. Resumen estadístico completo por materia")
        print("4. Ver aprobados y reprobados")
        print("5. Ver todos los estudiantes (JSON)")
        print("6. Promedio general por estudiante")
        print("7. Promedio general del grupo")
        print("0. Salir")

        opcion = input("Elige una opción: ")

        if opcion == "1":
            sistema.reporte_general()

        elif opcion == "2":
            print("\nVariables disponibles:\n1. edad\n2. tiempo_traslado_min\n3. horas_uso_mvil")
            campo = input("Elige el número de variable: ")

            if campo == "1":
                sistema.reporte_campo("edad")
            elif campo == "2":
                sistema.reporte_campo("tiempo_traslado_min")
            elif campo == "3":
                sistema.reporte_campo("horas_uso_mvil")
            else:
                print("Opción no válida.")

        elif opcion == "3":
            sistema.resumen_por_materia()

        elif opcion == "4":
            sistema.ver_aprobados_reprobados()

        elif opcion == "5":
            sistema.mostrar_estudiantes()

        elif opcion == "6":
            sistema.promedios_por_estudiante()

        elif opcion == "7":
            sistema.promedio_general_grupo()

        elif opcion == "0":
            print("Saliendo del sistema...")
            break

        else:
            print("Opción inválida, intenta de nuevo.")

# Ejecución del programa
if __name__ == "__main__":
    menu()
```
