```python
#librerias a importar
import tensorflow as tf
import pandas as pd
import numpy as np
```

```python
from IPython.utils import encoding
#leer datos
datos = pd.read_csv('/content/conversor.csv', sep=';', encoding='utf-8')
print(datos)
```

```python
c = datos["Celsius"]
f = datos["farenheits"]
print(c)
```

```python
#crear el modelo
modelo = tf.keras.Sequential()
modelo.add(tf.keras.layers.Dense(units=1, input_shape=[1]))
```

```python
#copilar el modelo
modelo.compile(optimizer=tf.keras.optimizers.Adam(0.1), loss='mean_squared_error')
```

```python
#entrenar el modelo
epocas =modelo.fit(c,f,epochs = 200, verbose=1)
```

```python
resp = modelo.predict(np.array([100.0]))
print(resp)
```
